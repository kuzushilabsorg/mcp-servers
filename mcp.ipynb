# What is MCP?

Model Context Protocol: A protocol for sharing context between models.

Task: Create an agent which can communicate with different apis - gmail, slack, database
- Need to build custom integrations with permissions and scoping control for it
- eg: Gmail docs may allow for deletion of messages but you do not want to give this functionality to your agent -> needs custom prompts or implementtion - guide the agent to do only the specific things you want to do
- This custom implmenentation - you can put in public and other people can use it - the same implementation for their agents
- Windsurf IDE can communicate with different apis - gmail, slack, database
- All implementations custom for Windsurf - now translate's to Cursor and other IDEs without MCP
- MCP = reduce complexity + reusability
- MCP = standard for anyone to connect their agent to your backend

## MCP Architecture
- MCP has a client-server architecture
- Hosts are LLM applications (Claude Desktop, IDES)  these initiate connections
- Clients maintain 1:1 connection with the server inside the Hosts
- Servers provide context, tools and prompts to the clients
- Tools: Enables LLMs to perform actions through the servers - executable functionality
- Prompts: Reusable prompt templates and workflows - predefined templates
